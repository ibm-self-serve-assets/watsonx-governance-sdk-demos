## Complete Workflow with Supervisory Agent

Now let's run the complete workflow orchestrated by the Supervisory Agent.

```python
from supervisory_agent import SupervisoryAgent

# Initialize Supervisory Agent
supervisor = SupervisoryAgent(
    apikey=os.getenv("WATSONX_APIKEY"),
    project_id=os.getenv("WATSONX_PROJECT_ID")
)

print("Supervisory Agent initialized successfully!")
```

### Scenario 1: Signed ESA from IBM

```python
# Define the scenario
seller_query = "I just received a signed ESA from IBM. What should I do next?"
contract_file_path = "docs/Confluent_IBM-1.30.2024.docx"
partner_name = "IBM"

print("SCENARIO: Signed ESA from IBM")
print("=" * 80)
print(f"Seller Query: {seller_query}")
print(f"Contract File: {contract_file_path}")
print(f"Partner: {partner_name}")
print("\nExecuting complete workflow...\n")
```

```python
# Run the complete workflow
result = supervisor.run(
    seller_query=seller_query,
    contract_file_path=contract_file_path,
    partner_name=partner_name
)

print("\n" + "=" * 80)
print("WORKFLOW COMPLETE")
print("=" * 80)
```

```python
# Display final result
print(result["final_result"])
```

## Inspect Individual Components

Let's examine the outputs from each agent in detail.

```python
import json

# Contract Summary
print("CONTRACT SUMMARY")
print("=" * 80)
if result.get("contract_summary"):
    structured = result["contract_summary"].get("structured_summary", {})
    print(json.dumps(structured, indent=2))
else:
    print("No contract summary available")
```

```python
# Partner Profile
print("PARTNER PROFILE")
print("=" * 80)
if result.get("partner_profile"):
    profile = result["partner_profile"]
    print(f"Partner: {profile.get('partner_name', 'Unknown')}")
    print(f"Maturity: {profile.get('maturity_level', 'Unknown')}")
    print(f"Sales Velocity: {profile.get('sales_velocity', 'Unknown')}")
    print(f"\nInternal Data:")
    print(json.dumps(profile.get('internal_data', {}), indent=2, default=str))
else:
    print("No partner profile available")
```

```python
# Action Recommendation
print("ACTION RECOMMENDATION")
print("=" * 80)
if result.get("action_recommendation"):
    action = result["action_recommendation"]
    
    # Risk Assessment
    risk = action.get("risk_assessment", {})
    print(f"Risk Level: {risk.get('risk_level', 'Unknown')}")
    print(f"Risk Score: {risk.get('risk_score', 0)}/100")
    print(f"\nRisk Factors:")
    for factor in risk.get('risk_factors', []):
        print(f"  - {factor}")
    
    # Recommended Action
    print(f"\nRecommended Action:")
    print(action.get("recommended_action", {}).get("raw_recommendation", "None"))
    
    # CRM Updates
    print(f"\nCRM Updates:")
    print(json.dumps(action.get("crm_updates", {}), indent=2))
else:
    print("No action recommendation available")
```
